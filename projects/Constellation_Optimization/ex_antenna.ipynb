```python
import numpy as np
```

```python
a = 9750
e = 0.6383
R_MOON = 1737.4  # [m] Radius of the Moon
tx_hpbw = np.arctan(R_MOON / (a * (1 - e)))

print(f"Transmitter antenna half-power beamwidth: {np.rad2deg(tx_hpbw):.2f} degrees")
```

```python
T_ant = 100  # [K] Antenna temperature
cable_loss_before_LNA = 1.0  # [dB] Cable loss before LNA
LNA_NF = 3.0  # [dB] LNA noise figure
LNA_gain = 30.0  # [dB] LNA gain
cable_loss_after_LNA = 10.0  # [dB] Cable loss after LNA
T0 = 290.0  # [K] Reference temperature

LNA_F = 10 ** (LNA_NF / 10)  # [linear] LNA noise figure
T_LNA = T0 * (LNA_F - 1)  # [K] LNA noise temperature
L1 = 10 ** (cable_loss_before_LNA / 10)  # [linear] Cable loss after LNA
T_cable1 = T0 * (L1 - 1)  # [K] Cable loss before LNA
T_anteff = L1 * T_ant  # [K] Antenna effective temperature
L2 = 10 ** (cable_loss_after_LNA / 10)  # [linear] Cable loss after LNA
T_cable2 = T0 * (L2 - 1)  # [K] Cable loss after LNA
G_LNA = 10 ** (LNA_gain / 10)  # [linear] LNA gain
T_sys = (
    T_anteff + T_cable1 + T_LNA / L1 + T_cable2 / (L1 * G_LNA)
)  # [K] System noise temperature

print(f"L1: {L1:.2f}")
print(f"T_cable1: {T_cable1:.2f} K")
print(f"T_anteff: {T_anteff:.2f} K")
print(f"T_LNA: {T_LNA:.2f} K")
print(f"L2: {L2:.2f}")
print(f"System noise temperature: {T_sys:.2f} K")
```

```python
import pylupnt as pnt

csv_file = "data/lnss_antenna.csv"

ant = pnt.Antenna("lunanet_esa.txt")

# read csv (1st column is elevation, 2nd column is gain)
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(csv_file, header=None)
elevations = df.iloc[:, 0].values  # elevation angles in degrees
gains = df.iloc[:, 1].values  # antenna gains in dB

print(elevations)

phi_ant = np.linspace(5, 85, 100)  # elevation angles in degrees
gains_ant = np.zeros_like(phi_ant)
phiv = ant.get_phi_vector()
thetav = ant.get_theta_vector()
gainv = ant.get_gain_matrix()
print("Phi vector:", phiv)
print("Theta vector:", thetav)
print("Gain vector:", gainv.T)
print("Antenna gains from CSV:", gains_ant)

plt.figure(figsize=(10, 6))
plt.plot(
    elevations,
    gains,
    marker="o",
    linestyle="-",
    color="b",
    label="Antenna Gain from CSV",
)
# plt.plot(phi_ant, gains_ant, marker='x', linestyle='--', color='r', label='Antenna Gain')
plt.plot()
plt.title("Antenna Gain vs Elevation Angle")
plt.xlabel("Elevation Angle (degrees)")
plt.ylabel("Antenna Gain (dB)")
plt.grid(True)
plt.legend()
plt.xlim(0, 90)
```
