```python
%load_ext autoreload
%autoreload 2

from src.failure_model import NavSatFailureModel
import numpy as np
import matplotlib.pyplot as plt

# ---- Example usage ----
model_short = NavSatFailureModel(
    tau1=0.25,  # years (≈3 months)
    S_tau1=0.95,  # survival at tau1
    beta0=0.6,  # infant mortality shape
    lam1=0.008,  # /year, nominal constant hazard
    tau2=5.0,  # start of wear-out (years)
    targets_years=(6.0, 8.0),
    targets_survival=(0.8, 0.5),  # survival at 6 and 8 years
)  # defaults described above

model_long = NavSatFailureModel(
    tau1=0.25,
    S_tau1=0.99,
    beta0=0.6,
    lam1=0.008,
    tau2=10.0,
    targets_years=(12.0, 15.0),
    targets_survival=(0.87, 0.75),
)

# ---- Plotting survival function ----
t = np.linspace(0, 15, 100)
dt = t[1] - t[0]
Slong = model_long.survival(t)
Sshort = model_short.survival(t)

fig, axes = plt.subplots(1, 2, figsize=(12, 3))

# plot 1: failure rate
ax = axes[0]
ax.plot(t, model_long.cond_fail_prob(t, 1.0), label="Failure Rate (Long)", color="blue")
ax.plot(
    t, model_short.cond_fail_prob(t, 1.0), label="Failure Rate (Short)", color="orange"
)
ax.set_title("Failure Rate")
ax.set_xlabel("Age (years)")
ax.set_ylabel("Failure Rate (/year)")
ax.grid()
ax.axvline(1.0, color="red", linestyle="--", label="Stage 1->1, 2->2, 3->3")
ax.axvline(6.0, color="green", linestyle="--", label="Stage 1->2, 2->3")
ax.axvline(11.0, color="purple", linestyle="--", label="Stage 1->3")
ax.set_xticks(np.arange(0, 16, 1))
ax.set_yticks(np.arange(0, 0.5, 0.05))
ax.set_xlim(0, 15)
ax.set_ylim(0, 0.5)
ax.legend()

# plot 2: survival function
ax = axes[1]
ax.plot(t, Slong, label="Survival Function (Long)", color="blue")
ax.plot(t, Sshort, label="Survival Function (Short)", color="orange")
ax.set_title("Survival Function")
ax.set_xlabel("Age (years)")
ax.set_ylabel("Survival Probability")
ax.grid()
ax.axvline(1.0, color="red", linestyle="--", label="Stage 1->1, 2->2, 3->3")
ax.axvline(6.0, color="green", linestyle="--", label="Stage 1->2, 2->3")
ax.axvline(11.0, color="purple", linestyle="--", label="Stage 1->3")
ax.set_xticks(np.arange(0, 16, 1))
ax.set_yticks(np.arange(0, 1.1, 0.1))
ax.set_xlim(0, 15)
ax.legend()

plt.tight_layout()
plt.savefig("figs/survival_function.pdf", dpi=300)
plt.show()
```
