# Process the Data from Moon TEC Simulation

```python
import pandas as pd
import os

DATA_DIR = "../../output/moon"
OUTPUT_DIR = "../../output/moon"

input_csv = os.path.join(DATA_DIR, "tec_gps_moon.csv")
df = pd.read_csv(input_csv)

# headers: "Time [s],PRN,Angle [deg],MinAlt [km],TEC [m^-2],Delay [m]"
df["Time"] = df["Time [s]"].astype(int)
df["PRN"] = df["PRN"].astype(int)
df["Angle"] = df["Angle [deg]"]
df["MinAlt"] = df["MinAlt [km]"]
df["TEC"] = df["TEC [m^-2]"]
df["Delay"] = df["Delay [m]"]
```

```python
# plot time series for different PRN
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize

fig, ax = plt.subplots()
# set font size
plt.rcParams.update({"font.size": 12})

# Create a normalization object from 0 to 30
norm = Normalize(vmin=0, vmax=30)

prn_list = df["PRN"].unique()
prn_max = 31
prn_cmap = "viridis"
# get a list of colors from the colormap, where min is 0 and max is prn_max
cmap = plt.get_cmap(prn_cmap)
prn_colors = [cmap(norm(i)) for i in range(prn_max)]
print(prn_colors)

for prn in prn_list:
    df_prn = df[df["PRN"] == prn]
    ax.scatter(df_prn["Time"] / 3600, df_prn["Delay"], color=prn_colors[prn])
plt.xlabel("Time [hr]")
plt.ylabel("Delay [m]")
plt.grid(True)
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, ticks=range(0, prn_max, 5), label="PRN")
# plt.title('Delay vs Time')
plt.savefig(os.path.join(OUTPUT_DIR, "delay_time.pdf"))
plt.show()
```

```python
# Min altitude vs Delay
fig, ax = plt.subplots()
# set fontsize to 12
ax.tick_params(axis="both", which="major", labelsize=12)
ax.tick_params(axis="both", which="minor", labelsize=12)
ax.set_xlabel("Min altitude [km]", fontsize=12)
ax.set_ylabel("Delay [m]", fontsize=12)
ax.grid(True)
for prn in prn_list:
    df_prn = df[df["PRN"] == prn]
    ax.scatter(df_prn["MinAlt"], df_prn["Delay"], color=prn_colors[prn])
# plt.title('Delay vs MinAlt')
# color bar
sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
sm.set_array([])
cbar = plt.colorbar(sm, ax=ax, ticks=range(0, prn_max, 5), label="PRN")
plt.savefig(os.path.join(OUTPUT_DIR, "delay_minalt.pdf"))
plt.show()
```
