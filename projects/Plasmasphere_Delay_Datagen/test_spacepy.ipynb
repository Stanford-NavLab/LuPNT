```python
import spacepy.coordinates as sc
from spacepy.time import Ticktock

# 1. Define Input: ECEF (ITRF) position and Time
# Example: 10,000 km altitude equatorial
x_ecef = -1000000.0  # meters
y_ecef = -4500000.0
z_ecef = 4000000.0
time_str = "2025-03-01T12:00:00"

# SpacePy expects distances in Earth Radii (Re) usually, but we can specify km/m if careful.
# Ideally, convert meters to radii for SpacePy:
Re = 6371000.0  # meters
pos_re = [x_ecef / Re, y_ecef / Re, z_ecef / Re]

# 2. Create the Coordinate Object
# 'GEO' is essentially ECEF
c_ecef = sc.Coords(pos_re, "GEO", "car")  # 'car' = Cartesian (x,y,z)
c_ecef.ticks = Ticktock([time_str], "ISO")

# 3. Convert to Solar Magnetic (SM)
c_sm = c_ecef.convert("SM", "sph")  # 'sph' = Spherical (radius, lat, lon)

# Access the data
# SpacePy spherical output is usually [Radius, Latitude, Longitude]
# Latitude is in degrees
sm_lat = c_sm.data[0][1]
sm_lon = c_sm.data[0][2]

print(f"SM Latitude: {sm_lat:.2f}°")
print(f"SM Longitude: {sm_lon:.2f}°")
```
