```python
%load_ext autoreload
%autoreload 2
import os
import pylupnt as pnt
import plotly.io as pio
import matplotlib.pyplot as plt

pio.renderers.default = "notebook"
```

```python
antenna = pnt.Antenna("Block-IIR-M_ACE")
fig = pnt.plot.plot_antenna_gain_pattern(antenna=antenna)
fig.show()
```

```python
# Find antenna_names as the filename (no extension) of iterating LUPNT_DATA_PATH/antenna and all with .txt extension
# Iterate directory and find all files with .txt extension
antenna_names = []
for root, dirs, files in os.walk(os.path.join(pnt.LUPNT_DATA_PATH, "antenna")):
    for file in files:
        if file.endswith(".txt"):
            antenna_names.append(file.split(".")[0])
print(antenna_names)
```

```python
m = 4
n = int(len(antenna_names) / m) + 1
fig = plt.figure(figsize=(4 * m, 4 * n))
axs = fig.subplots(n, m).flatten()
for i, name in enumerate(antenna_names):
    antenna = pnt.Antenna(name)
    pnt.plot.plot_antenna_gain_patter_2D(antenna=antenna, ax=axs[i])
    if i == 0:
        plt.legend()

plt.tight_layout()
plt.show()
```
